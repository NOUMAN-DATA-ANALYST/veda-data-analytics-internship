## **Getting started**
</div>

The document you are reading is not a static web page, but an interactive environment called a **Colab notebook** that lets you write and execute code.

For example, here is a **code cell** with a short Python script that computes a value, stores it in a variable, and prints the result:

```
seconds_in_a_day = 24 * 60 * 60
seconds_in_a_day
```

To execute the code in the above cell, select it with a click and then either press the play button to the left of the code, or use the keyboard shortcut "Command/Ctrl+Enter". To edit the code, just click the cell and start editing.

Variables that you define in one cell can later be used in other cells:

```
seconds_in_a_week = 7 * seconds_in_a_day
seconds_in_a_week
```

Colab notebooks allow you to combine **executable code** and **rich text** in a single document, along with **images**, **HTML**, **LaTeX** and more. ## Data science
</div>

With Colab you can harness the full power of popular Python libraries to analyze and visualize data. The code cell below uses **numpy** to generate some random data, and uses **matplotlib** to visualize it. To edit the code, just click the cell and start editing.

You can import your own data into Colab notebooks from your Google Drive account, including from spreadsheets, as well as from Github and many other sources. To learn more about importing data, and how Colab can be used for data science, see the links below under [Working with Data](#working-with-data).

```
import numpy as np
import IPython.display as display
from matplotlib import pyplot as plt
import io
import base64

ys = 200 + np.random.randn(100)
x = [x for x in range(len(ys))]

fig = plt.figure(figsize=(4, 3), facecolor='w')
plt.plot(x, ys, '-')
plt.fill_between(x, ys, 195, where=(ys > 195), facecolor='g', alpha=0.6)
plt.title("Sample Visualization", fontsize=10)

data = io.BytesIO()
plt.savefig(data)
image = F"data:image/png;base64,{base64.b64encode(data.getvalue()).decode()}"
alt = "Sample Visualization"
display.display(display.Markdown(F"""![{alt}]({image})"""))
plt.close(fig)
```

```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from google.colab import files

# Step 1: PC se direct Iris file upload karein
uploaded = files.upload()
```

```
# Data load karein
df = pd.read_csv('Iris.csv')

# First 5 rows check karein
print("--- FIRST 5 ROWS ---")
print(df.head())

# Datatypes aur missing values check karein
print("\n--- DATASET INFO ---")
df.info()

# Summary statistics (Mean, Std, Min, Max, Quantiles)
print("\n--- SUMMARY STATISTICS ---")
print(df.describe())
```

```
# Style set karein
sns.set_theme(style="whitegrid")

# --- CHART 1: Sepal Length Distribution (Histogram) ---
plt.figure(figsize=(7, 4))
sns.histplot(df['SepalLengthCm'], kde=True, color='teal')
plt.title('1. Distribution of Sepal Length')
plt.xlabel('Sepal Length (cm)')
plt.show()
print("Observation 1: Sepal Length mostly 5.0 cm se 6.5 cm ke darmayan hai aur normal bell-curve distribution dikhati hai.\n")

# --- CHART 2: Correlation Heatmap ---
plt.figure(figsize=(7, 5))
numeric_df = df.drop(columns=['Id', 'Species'], errors='ignore')
sns.heatmap(numeric_df.corr(), annot=True, cmap='coolwarm', fmt=".2f")
plt.title('2. Correlation Heatmap')
plt.show()
print("Observation 2: Petal Length aur Petal Width ke darmayan bohot strong positive correlation (0.96) hai.\n")

# --- CHART 3: Outlier Check (Boxplot) ---
plt.figure(figsize=(8, 5))
sns.boxplot(data=numeric_df)
plt.title('3. Outliers Check Across Features')
plt.show()
print("Observation 3: SepalWidthCm me kuch minor outliers hain, baaki features clean hain.\n")

# --- CHART 4: Pairplot by Species ---
sns.pairplot(df.drop(columns=['Id'], errors='ignore'), hue='Species', palette='Set1')
plt.suptitle('4. Feature Relationships Across Species', y=1.02)
plt.show()
print("Observation 4: Setosa species baaki do species se bilkul distinct aur alag group banati hai.\n")

# --- CHART 5: Petal Dimensions Scatter Plot ---
plt.figure(figsize=(7, 4))
sns.scatterplot(data=df, x='PetalLengthCm', y='PetalWidthCm', hue='Species', s=70)
plt.title('5. Petal Length vs Petal Width by Species')
plt.xlabel('Petal Length (cm)')
plt.ylabel('Petal Width (cm)')
plt.show()
print("Observation 5: Setosa ke petals sab se chote hain, Versicolor ke medium, aur Virginica ke petals sab se bade hain.")
```

### Top 3 Insights
1. **Clear Species Distinction:** Petal Length aur Petal Width ki measurements se *Iris-setosa* species ko baaki do species se 100% alag pehchana ja sakta hai.
2. **Strong Feature Correlation:** Petal Length aur Petal Width ke darmayan highly positive correlation (0.96) hai, jo linear relation show karta hai.
3. **Clean Dataset:** Dataset me Sepal Width ke ilawa koi noticeable outliers maujood nahi hain, jo data ki high quality ko darshata hai.

### Interview Questions & Answers

1. **How do you distinguish a genuine outlier from a data entry error?**
   * **Data Entry Error:** Yeh impossible ya extreme values hote hain (maslan Sepal length = 500 cm ya negative value). Yeh typing mistakes se hotay hain.
   * **Genuine Outlier:** Yeh natural variation ki वजह se hote hain lekin logical range ke andar hote hain (maslan exceptionally large natural flower sample).

2. **What's the difference between correlation and causation, and why does it matter in EDA?**
   * **Correlation:** Do variables ke ek sath badalne ko dikhata hai.
   * **Causation:** Dikhata hai ke ek variable doosre ko badalne par majboor kar raha hai.
   * **Why it matters:** EDA me correlation se sirf pattern milta hai, bina scientific domain test ke hum yeh nahi keh sakte ke ek variable doosre ka cause hai.

3. **Which chart type would you use to show a trend over 12 months, and why?**
   * **Line Chart (Line Plot):** Time-series data me 12 mahino ke continuous trend, seasonal changes, aur fluctuations ko dekhne ke liye Line Chart sab se best chart hai.
